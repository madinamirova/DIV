```python
class Car:
    def __init__(self, marka, model, il):
        self.marka = marka
        self.model = model
        self.il = il

    def show_info(self):
        print(f"Avtomobilin markası: {self.marka}, modeli: {self.model}, istehsal ili: {self.il}")


car1 = Car("BMW", "X5", 2022)
car2 = Car("Mercedes", "E-Class", 2023)
car3 = Car("Toyota", "Camry", 2021)

car1.show_info()
car2.show_info()
car3.show_info()
```

```python
# 2. Rectangle Class
# Rectangle adlı class yaradın. 
# Constructor vasitəsilə düzbucaqlının eni və uzunluğunu qəbul edin.
#  Sahəni və perimetri hesablayan iki ayrı metod yazın.

class Rectangle:
    def __init__(self, eni, uzunluq):
        self.eni = eni
        self.uzunluq = uzunluq

    def sahe(self):
        return self.eni * self.uzunluq

    def perimetr(self):
        return 2 * (self.eni + self.uzunluq)


rectangle1 = Rectangle(5, 10)

print("Sahə:", rectangle1.sahe())
print("Perimetr:", rectangle1.perimetr())
    
```

```python
#  Book Class
# Başlanğıc səviyyə
# Tapşırığın şərti:

# Book adlı class yaradın. Kitabın adı, müəllifi və səhifə sayı atributlarını təyin edin.
#  Kitab haqqında məlumatları göstərən metod yazın. Əlavə olaraq,
#  kitabın 300 səhifədən çox olub-olmadığını yoxlayan metod yaradın.

class Book:
    def __init__(self, ad, muellif, sehife_sayi):
        self.ad = ad
        self.muellif = muellif
        self.sehife_sayi = sehife_sayi

    def show_info(self):
        print(f"Kitabın adı: {self.ad}")
        print(f"Müəllifi: {self.muellif}")
        print(f"Səhifə sayı: {self.sehife_sayi}")

    def check_pages(self):
        if self.sehife_sayi > 300:
            print("Kitab 300 səhifədən çoxdur.")
        else:
            print("Kitab 300 səhifədən çox deyil.")


book1 = Book("1984", "George Orwell", 328)
book2 = Book("Balaca Şahzadə", "Antoine de Saint-Exupéry", 96)

book1.show_info()
book1.check_pages()

book2.show_info()
book2.check_pages()


```

```python
# 4. Employee Class
# Orta səviyyə
# Tapşırığın şərti:
# Employee adlı class yaradın.
#  İşçinin adı, vəzifəsi və aylıq maaşı constructor vasitəsilə təyin edilsin. 
# İşçinin maaşını müəyyən faizlə artıran metod yazın. 
# Həmçinin illik maaşı hesablayan metod yaradın.

class Employee:
    def __init__(self, ad, vezife, maas):
        self.ad = ad
        self.vezife = vezife
        self.maas = maas

    def maas_artir(self, faiz):
        self.maas = self.maas + (self.maas * faiz / 100)
        print(f"Yeni aylıq maaş: {self.maas} AZN")

    def illik_maas(self):
        return self.maas * 12


employee1 = Employee("Aysel", "Python Developer", 2000)

employee1.maas_artir(15)

print(f"İllik maaş: {employee1.illik_maas()} AZN")
```
