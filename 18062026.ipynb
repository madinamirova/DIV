```python
# 2. Rectangle Class
# Rectangle adlı class yaradın. 
# Constructor vasitəsilə düzbucaqlının eni və uzunluğunu qəbul edin.
#  Sahəni və perimetri hesablayan iki ayrı metod yazın.

class Rectangle:
    def __init__(self, eni, uzunluq):
        self.eni = eni
        self.uzunluq = uzunluq

    def sahe(self):
        return self.eni * self.uzunluq

    def perimetr(self):
        return 2 * (self.eni + self.uzunluq)


rectangle1 = Rectangle(5, 10)

print("Sahə:", rectangle1.sahe())
print("Perimetr:", rectangle1.perimetr())
    
```

```python
#  Book Class
# Başlanğıc səviyyə
# Tapşırığın şərti:

# Book adlı class yaradın. Kitabın adı, müəllifi və səhifə sayı atributlarını təyin edin.
#  Kitab haqqında məlumatları göstərən metod yazın. Əlavə olaraq,
#  kitabın 300 səhifədən çox olub-olmadığını yoxlayan metod yaradın.

class Book:
    def __init__(self, ad, muellif, sehife_sayi):
        self.ad = ad
        self.muellif = muellif
        self.sehife_sayi = sehife_sayi

    def show_info(self):
        print(f"Kitabın adı: {self.ad}")
        print(f"Müəllifi: {self.muellif}")
        print(f"Səhifə sayı: {self.sehife_sayi}")

    def check_pages(self):
        if self.sehife_sayi > 300:
            print("Kitab 300 səhifədən çoxdur.")
        else:
            print("Kitab 300 səhifədən çox deyil.")


book1 = Book("1984", "George Orwell", 328)
book2 = Book("Balaca Şahzadə", "Antoine de Saint-Exupéry", 96)

book1.show_info()
book1.check_pages()

book2.show_info()
book2.check_pages()


```

```python
# 4. Employee Class
# Orta səviyyə
# Tapşırığın şərti:
# Employee adlı class yaradın.
#  İşçinin adı, vəzifəsi və aylıq maaşı constructor vasitəsilə təyin edilsin. 
# İşçinin maaşını müəyyən faizlə artıran metod yazın. 
# Həmçinin illik maaşı hesablayan metod yaradın.

class Employee:
    def __init__(self, ad, vezife, maas):
        self.ad = ad
        self.vezife = vezife
        self.maas = maas

    def maas_artir(self, faiz):
        self.maas = self.maas + (self.maas * faiz / 100)
        print(f"Yeni aylıq maaş: {self.maas} AZN")

    def illik_maas(self):
        return self.maas * 12


employee1 = Employee("Aysel", "Python Developer", 2000)

employee1.maas_artir(15)

print(f"İllik maaş: {employee1.illik_maas()} AZN")
```

```python
class BankAccount:
    def __init__(self, owner, balance=0):
        self.owner = owner
        self.__balance = balance      # özəl (private) atribut

    def deposit(self, amount):
        if amount <= 0:
            print("Məbləğ müsbət olmalıdır.")
            return
        self.__balance += amount
        print(f"{amount} AZN hesaba əlavə olundu.")

    def withdraw(self, amount):
        if amount <= 0:
            print("Məbləğ müsbət olmalıdır.")
        elif amount > self.__balance:
            print(f"Balansda kifayət qədər vəsait yoxdur! Cari balans: {self.__balance} AZN")
        else:
            self.__balance -= amount
            print(f"{amount} AZN hesabdan çıxarıldı.")

    def get_balance(self):
        return self.__balance


account = BankAccount("Madina", 1000)
account.deposit(500)
account.withdraw(300)
account.withdraw(5000)
print(f"{account.owner} hesabının balansı: {account.get_balance()} AZN")
```

```python
# (x, y) koordinatlarını təmsil edən iki cütlük və 
# Lineilə başlanğıclaşdırılmış bir sinif yaradın .point1point2
# calculate_length()Xəttin uzunluğunu qaytaran bir metod əlavə edin . 
# İpucu: Məsafə düsturu -dır sqrt((x2-x1)^2 + (y2-y1)^2). 
# Sizə lazım ola bilər import math.
# calculate_slope()Xəttin mailliyini qaytaran bir metod əlavə edin . 
# İpucu: Maillik düsturu dır (y2-y1) / (x2-x1).

import math

class Line:
    def __init__(self, point1, point2):
        self.point1 = point1   # məsələn (1, 2)
        self.point2 = point2   # məsələn (4, 6)

    def calculate_length(self):
        x1, y1 = self.point1
        x2, y2 = self.point2
        return math.sqrt((x2 - x1) ** 2 + (y2 - y1) ** 2)

    def calculate_slope(self):
        x1, y1 = self.point1
        x2, y2 = self.point2
        if x2 == x1:
            return None   
        return (y2 - y1) / (x2 - x1)


# Yoxlama
line = Line((1, 2), (4, 6))
print(line.calculate_length())  
print(line.calculate_slope())   
```

```python
# Tapşırıq 8: Düzbucaqlı Sinif
# və Rectangleilə başlanğıclaşdırılmış bir sinif yaradın .width height
# ( ) calculate_area()sahəsini qaytaran bir metod əlavə edin .width * height
# calculate_perimeter()
# Perimetri ( 2 * (width + height)) qaytaran bir metod əlavə edin .

class Duzbucaqli:
    def __init__(self, width, height):
        self.width=width # en
        self.height= height # hundurluk

    def calculate_area (self):
            return self.width * self.height

    def calculate_perimetr(self):
            return 2* (self.width + self.height)

rect = Duzbucaqli(5,3)
print ("Sahesi:" , rect.calculate_area())
print ("Perimetri:" , rect.calculate_perimetr())
```

```python
# Silindr sinfi
# və Cylinderilə başlanğıclaşdırılmış bir sinif yaradın .radiusheight
# calculate_volume()Həcmi qaytaran bir metod əlavə edin . 
# İşarə: Həcm düsturu -dır pi * radius^2 * height. 
# İstifadə edə bilərsiniz math.pi.
# calculate_surface_area()Səth sahəsini qaytaran bir metod əlavə edin .
# İpucu: Səth sahəsi düsturu dır 2 * pi * radius * (radius + height).

import math

class Cylinder:
    def __init__(self, radius, height):
        self.radius = radius    
        self.height = height   

    def calculate_volume(self):
        return math.pi * self.radius ** 2 * self.height

    def calculate_surface_area(self):
        return 2 * math.pi * self.radius * (self.radius + self.height)


cyl = Cylinder(3, 5)
print("Həcmi:", round(cyl.calculate_volume(), 2))            
print("Səth sahəsi:", round(cyl.calculate_surface_area(), 2)) 

```

```python
class Kitab:
    def __init__(self, ad, muellif, sehife_sayi):
        self.ad = ad
        self.muellif = muellif
        self.sehife_sayi = sehife_sayi

    def __str__(self):
        return f"'{self.ad}' - {self.muellif}, {self.sehife_sayi} səhifə"

    def __len__(self):
        return self.sehife_sayi


# Sınaq
menim_kitabim = Kitab("Hobbit", "J.R.R. Tolkien", 310)

print(menim_kitabim)        
print(len(menim_kitabim))   
```
