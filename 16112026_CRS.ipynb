```python
#  Funksiyaları
# Məşq 1: Əsas Lambda
# Bir arqument götürüb kvadratını qaytaran lambda funksiyası yaradın.
# Onu adlı bir dəyişənə təyin edin squarer.
# Funksiyanı nömrə ilə çağırın 8və nəticəni çap edin.

squarer = lambda x: x ** 2
print(squarer(8))
```

```python
# Məhsulların qiymətlərindən ibarət list verilib. Hər məhsulun qiymətini 15% azaldın. 
# Endirimdən sonra qiyməti 50 AZN-dən aşağı olan məhsulları göstərin.

prices = [40, 80, 120, 200, 35, 150, 60]
discounted = map(lambda x: x * 0.85, prices)
filtered = filter(lambda x: x < 50, discounted)
result = list(filtered)
print(result)
```

```python
# İşçilərin maaşlarından ibarət list verilib. Maaşı 1500 AZN-dən az olan işçilərin maaşını 25% artırın.
# Digər işçilərin maaşlarını dəyişmədən saxlayın. Yeni maaşları bir listdə göstərin.

salaries = [900, 1200, 1800, 2500, 1400, 3000]
filtered_salaries = map(lambda x: x*1.25 if x<1500 else x, salaries)
result = list(filtered_salaries)
print(result)
```

```python
car = {
     "BMW": {"price": 45000, "year": 2020},
     "Mercedes": {"price": 60000, "year": 2022},
     "Toyota": {"price": 25000, "year": 2018},
     "Kia": {"price": 30000, "year": 2021},
     "Audi": {"price": 50000, "year": 2019},
    "Hyundai": {"price": 28000, "year": 2023}
 }

for c_name, c_info in car.items():
    if c_info["price"] < 30000:
        print(c_name, c_info)
 
# _______________________________________________________#
count = 0
for c_name, c_info in car.items():
    if c_info["price"] > 40000:
        count += 1

print("Qiymeti 40000 AZN-dən yuxarı olan car sayı:", count)
# _______________________________________________________#

#sorted_cars = QALIB    BUNA BAXxxxxxxxx
```

```python
# Sözlərin uzunluğuna görə sıralanması
words = [
    "computer", "AI", "programming", "data",
    "machine", "learning", "python", "code"
]
for word in sorted(words, key=lambda x: len(x)):
    print(word)
```

```python
# Məhsulların adı, qiyməti və stok sayı verilib. Stok sayı 10-dan az olan məhsulları tapın və onların qiymətlərini 15% artırın.
products = {
    "Laptop": {"price": 1500, "stock": 5},
    "Mouse": {"price": 30, "stock": 20},
    "Keyboard": {"price": 80, "stock": 7},
    "Monitor": {"price": 400, "stock": 4},
    "USB": {"price": 15, "stock": 30}
}
for p_name, p_info in products.items():
    if p_info["stock"] < 10:
        p_info["price"] *= 1.15
        print(p_name, p_info)
```
