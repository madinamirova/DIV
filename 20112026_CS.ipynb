```python
class BankAccount:
    def __init__(self, hesab_sahibinin_adi, balans):
        self.hesab_sahibinin_adi = hesab_sahibinin_adi
        self.balans = balans


class SavingsAccount(BankAccount):
    def __init__(self, hesab_sahibinin_adi, balans):
        super().__init__(hesab_sahibinin_adi, balans)

    def calculate_interest(self):
        return self.balans * 0.05


hesab = SavingsAccount("Madina", 2000)
faiz = hesab.calculate_interest()
print(f"{hesab.hesab_sahibinin_adi} üçün faiz məbləği: {faiz}")
```

```python
# 2. Polymorphism (Polimorfizm) — Ödəniş sistemləri Tapşırıq: 
# CreditCard və PayPal adlı iki class yaradın. Hər ikisində pay(amount) metodu olsun. 
# CreditCard ödənişə 2%, PayPal isə 3% komissiya əlavə etsin. 
# Hər iki class-dan obyektləri bir list-ə yığın və dövr vasitəsilə eyni metod çağırışı ilə ödənişləri həyata keçirin.
#  Məqsəd: Eyni metodun fərqli obyektlərdə fərqli davranmasını nümayiş etdirin.


class CreditCard:
    def pay(self, amount):
        komissiya = amount * 0.02
        total = amount + komissiya
        print(f"Kreditkart: {amount} AZN + {komissiya} komissiya = {total} AZN ödənildi")
        return total


class PayPal:
    def pay(self, amount):
        komissiya = amount * 0.03
        total = amount + komissiya
        print(f"PayPal: {amount} AZN + {komissiya} komissiya = {total} AZN ödənildi")
        return total


odenis_usullari = [CreditCard(), PayPal(), CreditCard()]

for usul in odenis_usullari:
    cixarilan = usul.pay(500)
    print(f"Hesabdan {cixarilan} AZN çıxarıldı\n")
```

```python
# 5. Public / Protected / Private — İstifadəçi məlumatları Tapşırıq: 
# User class-ı yaradın. Class daxilində public username, protected _email və private __password attribute-ları olsun. 
# Şifrəni yoxlamaq üçün check_password(password) metodu yazın. 
# O, daxil edilən şifrənin saxlanılan şifrəyə uyğun olub-olmadığını qaytarsın. 
# Məqsəd: Bu üç adlandırma formasının fərqini kod üzərində yoxlayın.

class User:
    def __init__(self, username, email, password):
        self.username = username      
        self._email = email          
        self.__password = password    

        def check_password(self, password): return self.__password == password


u = User("madina", "madina@example.com", "12345")
print(u.username)
print(u._email)
#print(u.__password)
```

```python
# 6. __str__ — Məhsulun mətn təsviri Tapşırıq: 
# Product class-ı yaradın. Məhsulun adı, qiyməti və kateqoriyası attribute kimi saxlanılsın. 
# Class daxilində __str__() metodunu yazın.
# Product obyekti print() edildikdə məhsul haqqında oxunaqlı məlumat göstərilsin. 
# Yoxlama: Bir neçə Product obyekti yaradıb print() ilə ekrana çıxarın.

class Product:
    def __init__(self, mehsulunAdi, qiymeti, kateqoriyasi):
        self.mehsulun_adi = mehsulunAdi
        self.mehsulun_qiymeti = qiymeti
        self.kateqoriyasi = kateqoriyasi

    def __str__(self):
        return (f"Məhsul: {self.mehsulun_adi} | "
                f"Qiymət: {self.mehsulun_qiymeti} AZN | "
                f"Kateqoriya: {self.kateqoriyasi}")

p1 = Product("Telefon", 3000, "Elektronika")
p2 = Product("Çay", 4.5, "Qida")
p3 = Product("Köynək", 3, "Geyim")
print(p1)
print(p2)
print(p3)

        
```

```python
# 7. __len__ — Kitabxana Tapşırıq: 
# Library class-ı yaradın. 
# Kitabların adlarını list-də saxlayın və yeni kitab əlavə etmək üçün add_book(name) metodu yazın. 
# __len__() metodunu elə reallaşdırın ki, len(library) çağırışı kitabxanadakı kitabların sayını qaytarsın.
class Library:
    def __init__(self):
        self.kitablarin_listi = []

    def add_book(self, name):
        self.kitablarin_listi.append(name)

    def __len__(self):
        return len(self.kitablarin_listi)


library = Library()
library.add_book("Martin Eden - Jack London")
library.add_book("1984 - George Orwell")

print(len(library))  
```

```python
# 8. __del__ — Server bağlantısı Tapşırıq: ServerConnection class-ı yaradın. 
# Obyekt yaradıldıqda server bağlantısının qurulduğunu bildirən mesaj göstərilsin.
#  __del__() metodu obyekt silinərkən bağlantının 
#  bağlandığını bildirən mesaj göstərsin. 
#  Obyekt yaradın və onu silmək üçün del istifadə edin. 
#  Qeyd: Python-da __del__ metodunun nə vaxt çağırılacağına həmişə zəmanət verilmir.

class ServerConnection:
    def __init__(self, server_name):
        self.server_name = server_name
        print(f"{self.server_name} serverinə bağlantı quruldu.")

    def __del__(self):
        print(f"{self.server_name} serveri ilə bağlantı bağlandı.")

connection = ServerConnection("main-server")
del connection
```

```python
# 9. @staticmethod — Kalkulyator 
# Tapşırıq: Calculator class-ı yaradın. 
# İçərisində is_even(number) adlı static method olsun. 
# Metod verilən ədədin cüt olub-olmadığını yoxlayaraq True və ya False qaytarsın.
#  Metodu class adı vasitəsilə çağırın.
# 9. @staticmethod — Kalkulyator
class Calculator:
    @staticmethod
    def is_even(number):
        return number % 2 == 0


print(Calculator.is_even(4))   # True
print(Calculator.is_even(7))   # False
```

```python
# 10. @classmethod — Universitet məlumatı Tapşırıq: 
# Student class-ı yaradın. Universitetin adı class attribute kimi saxlanılsın. 
# Universitetin adını dəyişmək üçün 
# change_university(cls, new_name) adlı class method yazın. 
# Metod çağırıldıqdan sonra yaradılmış tələbə obyektlərinin 
# universitet məlumatının yeniləndiyini yoxlayın.

class Student:
    university = "ADA University"   

    def __init__(self, name):
        self.name = name

    @classmethod
    def change_university(cls, new_name):
        cls.university = new_name


s1 = Student("Aysel")
s2 = Student("Murad")

print(s1.name, "-", s1.university)   
print(s2.name, "-", s2.university)   

Student.change_university("BDU")

print(s1.name, "-", s1.university)  
print(s2.name, "-", s2.university)   
```

```python
# 11. __init__ və super() — İşçi qeydiyyatı Tapşırıq: 
# Person class-ı yaradın. 
# Ad və yaş məlumatlarını __init__ vasitəsilə qəbul etsin. 
# Employee class-ı Person-dan miras alsın və əlavə olaraq
#  maaş məlumatını qəbul etsin. Employee daxilində 
# super() istifadə edərək ana class-ın __init__ metodunu çağırın. 
# İşçinin bütün məlumatlarını ekrana çıxarın.

class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age

class Employee(Person):
    def __init__(self, name, age, salary):
        super().__init__(name, age)  
        self.salary = salary

    def show_info(self):
        print(f"Ad: {self.name}")
        print(f"Yaş: {self.age}")
        print(f"Maaş: {self.salary} AZN")

emp = Employee("Leyla", 28, 1500)
emp.show_info()
```
