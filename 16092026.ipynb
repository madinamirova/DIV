```python
# Tələbələrin nəticələrini saxlayan kiçik analiz sistemi hazırlayın. Hər tələbə üçün ad, fənlər üzrə
# ballar və davamiyyət faizi saxlanılmalıdır. Proqram menyu vasitəsilə idarə olunmalı və istifadəçi
# seçim etdikcə uyğun əməliyyatı yerinə yetirməlidir.
# • Dataset üzrə ümumi orta balı hesablayın.
# • Ən yüksək orta bala malik tələbəni tapın.
# • Orta balı 80-dən yuxarı olan tələbələri göstərin.
# • Davamiyyəti 75%-dən aşağı olan tələbələri riskli kimi qeyd edin.
# • Tələbələri orta bala görə azalan sıra ilə sıralayın.
# • Hər fənn üzrə orta balı hesablayın.
# • Yeni tələbə əlavə etmək imkanı yaradın.
# • Mövcud tələbənin məlumatlarını yeniləmək imkanı yaradın.
# • Dictionary-dən istifadə edərək ümumi statistik məlumatları göstərin.
# • Proqramı while dövrü ilə menyu əsasında idarə edin; if/elif/else, def və lambda istifadə edin.
# • Boş dataset, mövcud olmayan tələbə və yanlış istifadəçi girişləri kimi halları nəzərə alın


telebeler = {
    "Ayan": {
        "ballar": {"Riyaziyyat": 85, "Python": 92, "AI": 88},
        "davamiyyet_faizi": 95
    },
    "Murad": {
        "ballar": {"Riyaziyyat": 70, "Python": 75, "AI": 68},
        "davamiyyet_faizi": 70
    },
    "Leyla": {
        "ballar": {"Riyaziyyat": 95, "Python": 90, "AI": 96},
        "davamiyyet_faizi": 88
    }
}


# 1. Köməkçi funksiya
def telebe_orta_balini_hesabla(fenn_ballari):
    toplam_bal = 0
    fenn_sayi = 0

    for bal in fenn_ballari.values():
        toplam_bal += bal
        fenn_sayi += 1

    if fenn_sayi > 0:
        return toplam_bal / fenn_sayi

    return 0


# 2. Hər tələbənin orta balı
print("=== HƏR TƏLƏBƏNİN ORTA BALI ===")

umumi_cem_orta = 0
aktiv_telebe_sayi = 0
en_yuksek_bal = 0
en_yaxsi_telebe = ""
siralanmis_siyahi = []

for ad, melumat in telebeler.items():
    fenn_ballari = melumat["ballar"]
    t_orta = telebe_orta_balini_hesabla(fenn_ballari)

    umumi_cem_orta += t_orta
    aktiv_telebe_sayi += 1

    print(f"{ad} - Orta balı: {t_orta:.2f}")

    siralanmis_siyahi.append((ad, t_orta))

    if t_orta > en_yuksek_bal:
        en_yuksek_bal = t_orta
        en_yaxsi_telebe = ad


# 3. Dataset üzrə ümumi orta bal
print("\n=== ÜMUMİ DATASET ÜZRƏ ORTA BAL ===")

if aktiv_telebe_sayi > 0:
    dataset_uzre_orta = umumi_cem_orta / aktiv_telebe_sayi
    print(f"Dataset üzrə ümumi orta bal: {dataset_uzre_orta:.2f}")


# 4. Ən yüksək orta bala malik tələbə
print("\n=== ƏN YÜKSƏK ORTA BALA MALİK TƏLƏBƏ ===")

print(f"Tələbə: {en_yaxsi_telebe} ({en_yuksek_bal:.2f})")


# 5. Orta balı 80-dən yuxarı olan tələbələr
print("\n=== ORTA BALI 80-DƏN YUXARI OLAN TƏLƏBƏLƏR ===")

for ad, melumat in telebeler.items():
    t_orta = telebe_orta_balini_hesabla(melumat["ballar"])

    if t_orta > 80:
        print(f"Uğurlu: {ad} ({t_orta:.2f})")


# 6. Davamiyyət analizi
print("\n=== DAVAMİYYƏT ANALİZİ ===")

for ad, melumat in telebeler.items():
    davamiyyet = melumat["davamiyyet_faizi"]

    if davamiyyet >= 75:
        print(f"{ad} - Davamiyyəti yaxşıdır: {davamiyyet}%")
    else:
        print(f"{ad} - RİSKLİDİR (Davamiyyəti azdır): {davamiyyet}%")


# 7. Tələbələrin orta bala görə sıralanması
print("\n=== TƏLƏBƏLƏRİN ORTA BALA GÖRƏ AZALAN SIRA İLƏ SIRALANMASI ===")

siralanmis_siyahi.sort(key=lambda x: x[1], reverse=True)

for sira, (ad, orta) in enumerate(siralanmis_siyahi, 1):
    print(f"{sira}. {ad} - Orta bal: {orta:.2f}")


# 8. Hər fənn üzrə orta bal
print("\n=== HƏR FƏNN ÜZRƏ ORTA BAL ===")

fenn_toplam_ballari = {}
fenn_sayaclari = {}

for melumat in telebeler.values():
    fenn_ballari = melumat["ballar"]

    for fenn, bal in fenn_ballari.items():

        if fenn not in fenn_toplam_ballari:
            fenn_toplam_ballari[fenn] = 0
            fenn_sayaclari[fenn] = 0

        fenn_toplam_ballari[fenn] += bal
        fenn_sayaclari[fenn] += 1


for fenn in fenn_toplam_ballari:
    fenn_orta = fenn_toplam_ballari[fenn] / fenn_sayaclari[fenn]
    print(f"{fenn} üzrə orta bal: {fenn_orta:.2f}")
```

```python

# Yeni tələbə əlavə etmək imkanı

def yeni_telebe_elave_et():
    ad = input("Yeni tələbənin adını daxil edin: ")

    if ad in telebeler:
        print(f"{ad} adlı tələbə artıq mövcuddur.")

    else:
        ballar = {}

        davamiyyet_faizi = float(
            input("Davamiyyət faizini daxil edin: ")
        )

        fenn_sayi = int(input("Tələbənin neçə fənni var? "))

        for i in range(fenn_sayi):
            fenn = input(f"{i+1}. fənnin adını daxil edin: ")

            bal = float(input(f"{fenn} üzrə balı daxil edin: "))

            ballar[fenn] = bal

        telebeler[ad] = {
            "ballar": ballar,
            "davamiyyet_faizi": davamiyyet_faizi
        }

        print(f"{ad} adlı tələbə əlavə edildi.")


yeni_telebe_elave_et()
```

```python
#  Mövcud tələbənin məlumatlarını yeniləmək imkanı yaradın.

def movcud_telebe_melumatlarini_yenile():
    ad = input("Mövcud tələbənin adını daxil edin: ")

    if ad not in telebeler:
        print(f"{ad} adlı tələbə mövcud deyil.")

    else:
        melumat = telebeler[ad]

        print(f"{ad} adlı tələbənin məlumatları:")
        print(f"Ballar: {melumat['ballar']}")
        print(f"Davamiyyət faizi: {melumat['davamiyyet_faizi']}")

        yeni_davamiyyet = float(
            input("Yeni davamiyyət faizini daxil edin (0-100): ")
        )

        melumat["davamiyyet_faizi"] = yeni_davamiyyet

        fenn_sayi = int(input("Tələbənin neçə fənni var? "))

        for i in range(fenn_sayi):
            fenn = input(f"{i+1}. fənnin adını daxil edin: ")

            bal = float(input(f"{fenn} üzrə yeni balı daxil edin: "))

            melumat["ballar"][fenn] = bal

        print(f"{ad} adlı tələbənin məlumatları yeniləndi.")
```
