```python

# Restoran sifariş sistemi 
# Tapşırığın şərtləri: 
# 1. Menyu yaratmaq üçün dictionary (lüğət) istifadə et. Menyuda yeməklərin adları və qiymətləri olsun. 
# 2. Müştərinin sifariş etdiyi yeməkləri saxlamaq üçün siyahı (list) yarat. 
# 3. Müştərinin sifariş siyahısındakı yeməklərin üzərindən for dövrü ilə keç. 
# 4. Hər yeməyin qiymətini menyudan götürərək ümumi məbləği (total) hesabla. 
# 5. Əgər ümumi məbləğ 100 manatdan çoxdursa, 20% endirim tətbiq et (if). 
# 6. Əgər ümumi məbləğ 50 manatdan çoxdursa, 10% endirim tətbiq et (elif). 
# 7. Əks halda, heç bir endirim tətbiq etmə (else). 
# 8. Sonda müştəriyə çek təqdim et. Çekdə aşağıdakılar göstərilsin: 
#  - Sifariş edilən yeməklərin adları. 
#  - Hər yeməyin qiyməti. 
#  - Endirim faizi. 
#  - Endirim məbləği. 
#  - Yekun ödəniləcək məbləğ. 


menu = {
    "Pizza": 25,
    "Burger": 15,
    "Salat": 10,
    "Steyk": 45,
    "Pasta": 20,
    "Sushi": 35
}

sifaris = ["Pizza", "Burger", "Steyk"]

total = 0

print("========== RESTORAN ÇEKİ ==========")

for yemek in sifaris:
    qiymet = menu[yemek]
    total += qiymet
    print(yemek, "-", qiymet, "AZN")

print("-----------------------------------")
print("Ümumi məbləğ:", total, "AZN")

if total > 100:
    endirim_faizi = 20
elif total > 50:
    endirim_faizi = 10
else:
    endirim_faizi = 0

endirim_meblegi = total * endirim_faizi / 100
yekun_mebleg = total - endirim_meblegi

print("Endirim faizi:", endirim_faizi, "%")
print("Endirim məbləği:", endirim_meblegi, "AZN")
print("Yekun ödəniş:", yekun_mebleg, "AZN")
print("===================================")
```

```python
# Tapşırıq 2 — Siyahı statistikası

# Məqsəd: list, for, sum/max/min/len, def ilə funksiya yazmaq.

# [3, 8, 1, 9, 4]
#       │
#       ├─ sum → 25      ├─ max → 9
#       ├─ min → 1       └─ len → 5  → orta = 25/5
#       │
#       └─ for ──▶ cüt ədədlər: 8, 4 → say = 2

# Şərtlər:

# statistika(ededler) adlı funksiya yaz — parametr bir siyahıdır.
# Funksiya cəmi, ən böyüyü, ən kiçiyi və ortanı ekrana çıxarsın (sum, max, min, len istifadə et).
# for dövrü ilə yalnız cüt ədədləri say (% operatoru ilə) — bu hissədə hazır funksiya yox, özün say.
# Funksiyanı ən azı 2 fərqli siyahı ilə çağır.
# Faylın sonuna çağırışları əlavə et, nəticəni göndər.

def statistika(ededler):
    cem = sum(ededler)
    en_boyuk = max(ededler)
    en_kicik = min(ededler)
    orta = cem / len(ededler)

    cut_say = 0
    for eded in ededler:
        if eded % 2 == 0:
            cut_say += 1

    print("Cəm:", cem)
    print("Ən böyük:", en_boyuk)
    print("Ən kiçik:", en_kicik)
    print("Orta:", orta)
    print("Cüt ədədlərin sayı:", cut_say)

statistika([10, 20, 35])
```
