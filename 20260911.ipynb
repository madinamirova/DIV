```python
def EnUzunSozUzunlugu(cumle):
    sozler = cumle.split()

    en_uzun_soz = max(sozler, key=len)
    return len(en_uzun_soz)

result = EnUzunSozUzunlugu("salam men Madina hg ghcygfgh hjgfyg")
print(result)  
```

```python

# İki tam ədədin ƏBOB-unu Evklid alqoritmi ilə hesablayan proqram yazın. 
# Həll zamanı while dövründən istifadə edin. 
# math.gcd funksiyasından istifadə etmək olmaz. Mənfi ədədlər və sıfır daxil edilə biləcəyini nəzərə alın.

input1 = int(input("Birinci tam ededi daxil et"))
input2 = int(input("Ikinci tam ededi daxil et"))

while input2 != 0:
    eded = input2
    input2 = input1 % input2
    input1 = eded

print("ƏBOB:", input1)
```

```python

def eded_siyahisi(siyahi):
    teze_siyahi = []

    say = 1

    for i in range(1, len(siyahi)):
        if siyahi[i] == siyahi[i - 1]:
            say += 1
        else:
            teze_siyahi.append((siyahi[i - 1], say))
            say = 1

    if siyahi:
        teze_siyahi.append((siyahi[-1], say))

    return teze_siyahi


eded_siyahi = [1, 1, 1, 2, 2, 5, 1, 1]

result = eded_siyahisi(eded_siyahi)

print(result)
```
