```python
tweet = "  WOW, what a great game! #sports #python  "

# 1. Kiçik hərflər
tweet_lower = tweet.lower()

# 2. Boşluqları silmək
clean_tweet = tweet_lower.strip()

# 3. Sözü əvəz etmək
updated_tweet = clean_tweet.replace("#python", "#DataScience")

# 4. Sözlər siyahısına bölmək
words = updated_tweet.split()

print(tweet_lower)
print(clean_tweet)
print(updated_tweet)
print(words)
```
