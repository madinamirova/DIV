```python
#Şərtlər:

#Boş dict yarat — açar: məhsul adı, dəyər: qiymət.
#while dövrü ilə istifadəçidən məhsul adı soruş; bitdi yazanda dövr dayansın (break).
#Hər məhsul üçün qiyməti də soruş (float) və dict-ə əlavə et.
#Sonda for ilə bütün məhsulları ad — qiymət formatında çap et.
#Ümumi məbləği sum ilə hesabla (ipucu: sebet.values()), f-string ilə .2f formatında göstər.



mehsullar = {}

while True:
    mehsul_adi = input("Məhsul adını daxil edin (bitdi yazmaq üçün 'bitdi' yazın): ")
    if mehsul_adi.lower() == 'bitdi':
        break
    try:
        qiymet = float(input(f"{mehsul_adi} üçün qiyməti daxil edin: "))
        mehsullar[mehsul_adi] = qiymet
    except ValueError:
        print("Zəhmət olmasa düzgün bir qiymət daxil edin.")
```

```python
# Tapşırıq 3 — Test nəticələri hesabatı

# Məqsəd: dict, for, if, def, len — real QA ssenarisində.

# Test run-dan belə nəticə gəlib:

# python
# testler = {
#     "login": "PASS",
#     "odenis": "FAIL",
#     "profil": "PASS",
#     "axtaris": "FAIL",
#     "cixis": "PASS",
# }

# Şərtlər:

# hesabat(testler) funksiyası yaz.
# for ilə dict-i gəz: PASS və FAIL saylarını hesabla.
# FAIL olan testlərin adlarını ayrıca siyahıya yığ.
# Çap et (f-string): ümumi sayı, pass faizi (.1f), və fail siyahısını.
testler = {
    "login": "PASS",
    "odenis": "FAIL",
    "profil": "PASS",
    "axtaris": "FAIL",
    "cixis": "PASS",
}

def hesabat(testler):
    failler = []
    pass_sayi = 0
    fail_sayi = 0

    for test_adi, netice in testler.items():

        if netice == "PASS":
            pass_sayi += 1

        elif netice == "FAIL":
            fail_sayi += 1
            failler.append(test_adi)

    umumi_sayi = len(testler)
    pass_faizi = (pass_sayi / umumi_sayi) * 100

    print(f"Ümumi test sayı: {umumi_sayi}")
    print(f"PASS sayı: {pass_sayi}")
    print(f"FAIL sayı: {fail_sayi}")
    print(f"PASS faizi: {pass_faizi:.1f}%")
    print(f"Fail olan testlər: {failler}")


hesabat(testler)
```
