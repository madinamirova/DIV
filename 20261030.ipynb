```python
# Tapşırıq 1: Siyahının Yaradılması və Girişi
# colors"Qırmızı", "yaşıl", "mavi", "yellow" və "bənövşəyi" sətirlərini ehtiva edən bir siyahı yaradın .
# Siyahıdakı üçüncü elementi çap edin.
# Mənfi indeksləmədən istifadə edərək, siyahıdakı sonuncu elementi çap edin.
# İlk üç rəngi ("qırmızı", "yaşıl", "mavi") əldə etmək üçün primary_colorsdilimləyərək 
# yeni bir siyahı yaradın . Yeni siyahını çap edin.colors



colors = ["Qırmızı", "yaşıl", "mavi", "yellow", "bənövşəyi"]

print("\n--- TAPŞIRIQ 1 ---")
print("Üçüncü element:", colors[2])
print("Sonuncu element:", colors[-1])

primary_colors = colors[:3]
print("İlk üç rəng:", primary_colors)
```

```python

print("\n--- TAPŞIRIQ 5 ---")

def grade_calculator(score):
    if 90 <= score <= 100:
        return "A"
    elif 80 <= score <= 89:
        return "B"
    elif 70 <= score <= 79:
        return "C"
    elif 60 <= score <= 69:
        return "D"
    elif 0 <= score < 60:
        return "F"
    else:
        return "Yanlış qiymət"

scores = [95, 82, 78, 61, 45]

for score in scores:
    print(score, "->", grade_calculator(score))
```
